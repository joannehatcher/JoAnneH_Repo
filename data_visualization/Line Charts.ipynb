```python
df.head()
```

### Task 1: Trends in Compensation Over Age Groups


##### 1. Line Chart of Median `ConvertedCompYearly` by Age Group


- Track how the median yearly compensation (ConvertedCompYearly) changes across different age groups.

- Use a line chart to visualize these trends.


```python
comp_age = df.groupby("Age")["ConvertedCompYearly"].median().dropna()

comp_age.plot(kind= "line", marker = "o", figsize = (12,8) )
plt.xticks (rotation = 45, ha = "right")
plt.title("Median Yearly compensation by Age group")
plt.xlabel("Age Group")
plt.ylabel("Yearly Compensation")
plt.grid(True, axis = "y", linestyle = "--", alpha = 1, color = "red")
plt.tight_layout()
plt.show()
```

##### 2. Line Chart of Median `ConvertedCompYearly` for Ages 25 to 45


For a closer look, plot a line chart focusing on the median compensation for respondents between ages 25 and 45.


```python
df["Age"] = df["Age"].str.strip()
selected_age = df[df["Age"].isin(["25-34 years old","35-44 years old"])]
comp_selected_age_median = selected_age.groupby("Age")["ConvertedCompYearly"].median()
print(comp_selected_age_median)

comp_selected_age_median.plot(kind = "line", marker = "o", figsize = (10,6))
plt.grid(True, axis = "y", linestyle = "--")
plt.title("Median ConvertedCompYearly for Ages 25 to 45")
plt.xlabel("Age")
plt.ylabel("Median Yearly Compensation")
plt.xticks(rotation = 45, ha = "right")
plt.tight_layout()
plt.show()

print(comp_selected_age_median)
print(comp_selected_age_median.index)
```

```python
age_map = {"Under 18 years old":18,
          "35-44 years old":39.5,
          "45-54 years old":49.5,
          "18-24 years old":21,
          "25-34 years old":29.5,
          "55-64 years old":59.5,
          "65 years or older":65
           
}

df["Age"] = df["Age"].str.strip()
df["Age"] = df["Age"].map(age_map)
comp_selected_age = df[(df["Age"]>=25) & (df["Age"]<=45)]
comp_selected_age_median = comp_selected_age.groupby ("Age")["ConvertedCompYearly"].median().dropna()
print(comp_selected_age_median)


comp_selected_age_median.plot(kind = "line", marker = "o", figsize = (10,6))
plt.grid(True, axis = "y", linestyle = "--")
plt.title("Median ConvertedCompYearly for Ages 25 to 45")
plt.xlabel("Age")
plt.ylabel("Median Yearly Compensation")
plt.xticks(rotation = 45, ha = "right")
plt.tight_layout()
plt.show()
```

### Task 2: Trends in Job Satisfaction by Experience Level



##### 1. Line Chart of Job Satisfaction (`JobSatPoints_6`) by Experience Level



- Use a column that approximates experience level to analyze how job satisfaction changes with experience.

- If needed, substitute an available experience-related column for `Experience`.


```python
job_sat_exp =(df.groupby("WorkExp")["JobSatPoints_6"].mean().dropna())
job_sat_exp.plot(kind = "line", marker = "o", figsize = (10,8))
plt.title("Trends in Job Satisfaction by Experience Level")
plt.xlabel("Years of work Experience")
plt.ylabel ("Average Job Satisfaction score")
plt.grid(True, axis = "y", linestyle = "--")
plt.tight_layout()
plt.show()
```

### Task 3: Trends in Job Satisfaction and Compensation by Experience


##### 1.Line Chart of Median ConvertedCompYearly Over Experience Level

- This line chart will track how median compensation (`ConvertedCompYearly`) changes with increasing experience.

- Use a column such as `WorkExp` or another relevant experience-related column.


```python
sat_comp_exp =(df.groupby("WorkExp")["ConvertedCompYearly"].median().dropna())
sat_comp_exp.plot(kind = "line", marker = "o", figsize = (10,8))
plt.title("Median Yearly Compensation by Experience Level")
plt.xlabel("Years of Work Experience")
plt.ylabel ("Median Yearly Compensation")
plt.grid(True, axis = "y", linestyle = "--")
plt.tight_layout()
plt.show()
```

##### 2.Line Chart of Job Satisfaction (`JobSatPoints_6`) Across Experience Levels

- Create a line chart to explore trends in job satisfaction (`JobSatPoints_6`) based on experience level.

- This chart will provide insight into how satisfaction correlates with experience over time


```python
sat_exp =(df.groupby("WorkExp")["JobSatPoints_6"].median().dropna())
sat_exp.plot(kind = "line", marker = "o", figsize = (10,8))
plt.title("Median Job Satisfaction by Experience Level")
plt.xlabel("Years of Work Experience")
plt.ylabel ("Median Job Satisfaction")
plt.grid(True, axis = "y", linestyle = "--")
plt.tight_layout()
plt.show()
```
