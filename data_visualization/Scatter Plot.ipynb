## Objectives

- Use scatter plots to identify trends and patterns in the dataset.

- Focus on visualizations centered on scatter plots for better data-driven insights.


## Setup: Working with the Database



**Install and import the required libraries**


```python
!pip install pandas
!pip install matplotlib
!pip install seaborn

import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

#### Step 1: Load the dataset


```python
file_path = "https://cf-courses-data.s3.us.cloud-object-storage.appdomain.cloud/n01PQ9pSmiRX6520flujwQ/survey-data.csv"

df = pd.read_csv(file_path)
```


### Task 1: Exploring Relationships with Scatter Plots



#### 1. Scatter Plot for Age vs. Job Satisfaction



Visualize the relationship between respondents' age (`Age`) and job satisfaction (`JobSatPoints_6`). Use this plot to identify any patterns or trends.




```python
df.head()
```

```python
plt.figure(figsize=(10, 8))
plt.scatter(df["Age"], df["JobSatPoints_6"])
plt.xticks(rotation = 45, ha= "right")
plt.xlabel("Age")
plt.ylabel("Job Satisfaction")
plt.show()

age_map = {
    "Under 18 years old": 18,
    "18-24 years old": 21,
    "25-34 years old": 29.5,
    "35-44 years old": 39.5,
    "45-54 years old": 49.5,
    "55-64 years old": 59.5,
    "65 years or older": 65
}

df["Age_numeric"] = df["Age"].map(age_map)

plt.figure(figsize=(10, 8))
plt.scatter(df["Age_numeric"], df["JobSatPoints_6"])
plt.xticks(rotation = 45, ha= "right")
plt.xlabel("Age")
plt.ylabel("Job Satisfaction")
plt.show()
```

#### 2. Scatter Plot for Compensation vs. Job Satisfaction


Explore the relationship between yearly compensation (`ConvertedCompYearly`) and job satisfaction (`JobSatPoints_6`) using a scatter plot.


```python
plt.figure(figsize=(10,8))
plt.scatter(df["JobSatPoints_6"], df["ConvertedCompYearly"])
plt.xlabel("Job Satisfaction")
plt.ylabel("Yearly Compensation")
plt.show()
```

### Task 2: Enhancing Scatter Plots


#### 1. Scatter Plot with Trend Line for Age vs. Job Satisfaction



Add a regression line to the scatter plot of Age vs. JobSatPoints_6 to highlight trends in the data.


```python
plt.figure(figsize = (10,8))
sns.regplot(data = df, x="Age_numeric", y = "JobSatPoints_6")
plt.xlabel("Age")
plt.ylabel("Job Satisfaction")
plt.title("Trend Line for Age vs. Job Satisfaction")
plt.show()
```

#### 2. Scatter Plot for Age vs. Work Experience


Visualize the relationship between Age (`Age`) and Work Experience (`YearsCodePro`) using a scatter plot.
YearsCodePro

```python
df["YearsCodePro"] = pd.to_numeric(df["YearsCodePro"], errors = "coerce")

plt.figure(figsize = (10, 8))
plt.scatter(df["Age"], df["YearsCodePro"])
plt.title("Relationship between Age and Work Experience")
plt.xlabel("Age")
plt.ylabel("Work Experience")
plt.xticks(rotation = 45, ha="right")
plt.show()


plt.figure(figsize = (10, 8))
plt.scatter(df["Age_numeric"], df["YearsCodePro"])
plt.title("Relationship between Age and Work Experience")
plt.xlabel("Age")
plt.ylabel("Work Experience")
plt.xticks(rotation = 45, ha="right")
plt.show()
```

### Task 3: Combining Scatter Plots with Additional Features


#### 1. Bubble Plot of Compensation vs. Job Satisfaction with Age as Bubble Size



Create a bubble plot to explore the relationship between yearly compensation (`ConvertedCompYearly`) and job satisfaction (`JobSatPoints_6`), with bubble size representing age.


```python
plt.figure(figsize=(10, 8))
plt.scatter(df["JobSatPoints_6"], df["ConvertedCompYearly"], s = df["Age_numeric"]*3, alpha = 0.25, c = "red")
plt.title("Yearly compensation and job satisfaction with Age")
plt.xlabel("Job Satisfaction")
plt.ylabel("Yearly Compensation")
plt.yscale("log")
plt.ylim(0,1000000)
plt.grid(alpha = 0.75)
plt.show()
```

#### 2. Scatter Plot for Popular Programming Languages by Job Satisfaction


Visualize the popularity of programming languages (`LanguageHaveWorkedWith`) against job satisfaction using a scatter plot. Use points to represent satisfaction levels for each language.


```python
language_jobsat = df[["LanguageHaveWorkedWith", "JobSatPoints_6"]].copy()

language_jobsat ["JobSatPoints_6"] = pd.to_numeric(language_jobsat ["JobSatPoints_6"], errors = "coerce")

language_jobsat = language_jobsat.dropna(subset = ["LanguageHaveWorkedWith","JobSatPoints_6"])

language_jobsat["Languages"] = language_jobsat["LanguageHaveWorkedWith"].str.split(";")

language_jobsat = language_jobsat.explode("Languages")
language_jobsat["Languages"] = language_jobsat["Languages"].str.strip()

print(language_jobsat)


plt.figure(figsize= (10, 8))
plt.scatter(language_jobsat["Languages"], language_jobsat["JobSatPoints_6"])
plt.xlabel("Programming Language")
plt.ylabel("Job Satisfaction")
plt.title("Programming Languages vs Job Satisfaction")
plt.xticks(rotation = 45, ha = "right")
plt.show()


lang = df.copy()
lang["JobSatPoints_6"] = pd.to_numeric(lang["JobSatPoints_6"], errors= "coerce")
lang = lang.dropna(subset=["LanguageHaveWorkedWith", "JobSatPoints_6"])
lang["Languages"] = lang["LanguageHaveWorkedWith"].str.split(";")
lang = lang.explode("Languages")
lang["Languages"] = lang["Languages"].str.strip()


plt.figure(figsize= (10, 8))
plt.scatter(lang["Languages"], lang["JobSatPoints_6"], alpha= 0.3)
plt.xlabel("Programming Language")
plt.ylabel("Job Satisfaction")
plt.title("Programming Languages vs Job Satisfaction")
plt.xticks(rotation = 45, ha = "right")

plt.show()
```

### Task 4: Scatter Plot Comparisons Across Groups


#### 1. Scatter Plot for Compensation vs. Job Satisfaction by Employment Type


Visualize the relationship between yearly compensation (`ConvertedCompYearly`) and job satisfaction (`JobSatPoints_6`), categorized by employment type (`Employment`). Use color coding or markers to differentiate between employment types.


```python
comp_employ = df[["JobSatPoints_6","ConvertedCompYearly", "Employment"]].copy()
comp_employ["JobSatPoints_6"] = pd.to_numeric(comp_employ["JobSatPoints_6"], errors = "coerce")
comp_employ["ConvertedCompYearly"] = pd.to_numeric(comp_employ["ConvertedCompYearly"], errors = "coerce")
comp_employ= comp_employ.dropna()

comp_employ["employ_simple"] = comp_employ["Employment"].str.split(";").str[0]


plt.figure (figsize=(12,8))
sns.scatterplot(data= comp_employ, x= "JobSatPoints_6", y = "ConvertedCompYearly", hue = "employ_simple", alpha = .75)
plt.title("Compensation vs. Job Satisfaction by Employment Type")
plt.xlabel("Job Satisfaction")
plt.ylabel ("Yearly Compensation")
plt.ylim(0,500000)
plt.xticks(rotation = 45, ha= "right")
plt.show()
```

#### 2. Scatter Plot for Work Experience vs. Age Group by Country


Compare work experience (`YearsCodePro`) across different age groups (`Age`) and countries (`Country`). Use colors to represent different countries and markers for age groups.


```python
work_age = df[["YearsCodePro","Age", "Country"]].copy()
work_age["YearsCodePro"] = pd.to_numeric(work_age["YearsCodePro"], errors = "coerce")
work_age= work_age.dropna()


plt.figure (figsize=(12,8))
sns.scatterplot(data= work_age, x= "Age", y = "YearsCodePro", hue = "Country", alpha = .75)
plt.title("Work Experience vs. Age Group by Country")
plt.xlabel("Age")
plt.ylabel ("Years Experience")
plt.xticks(rotation = 45, ha= "right")
plt.show()
```
