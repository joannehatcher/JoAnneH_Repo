### Step 1: Import Libraries and Load Data


- Import the `pandas`, `matplotlib.pyplot`, and `seaborn` libraries.


- You will begin with loading the dataset. You can use the pyfetch method if working on JupyterLite. Otherwise, you can use pandas' read_csv() function directly on their local machines or cloud environments.


```python
# Import necessary libraries
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

# Load the Stack Overflow survey dataset
data_url = 'https://cf-courses-data.s3.us.cloud-object-storage.appdomain.cloud/n01PQ9pSmiRX6520flujwQ/survey-data.csv'
df = pd.read_csv(data_url)

# Display the first few rows of the dataset
df.head()
```

### Step 2: Examine the Structure of the Data


- Display the column names, data types, and summary information to understand the data structure.

- Objective: Gain insights into the dataset's shape and available variables.


```python
print("Column names: \n")
print(df.columns)
print("Data types \n")
print(df.dtypes)
print ("Info \n")
print (df.info())
print("SHAPE", df.shape)
```

### Step 3: Handle Missing Data


- Identify missing values in the dataset.

- Impute or remove missing values as necessary to ensure data completeness.



```python
missing_data = df.isnull().sum()
print("All columns:", missing_data)
missing_data = missing_data[missing_data>0]
print("\nMissing columns:",missing_data)
print("\nMissing sum:", missing_data.sum())
```

```python

df_clean = df.dropna()
print (df.shape)
print(df_clean.shape)

print(df.isna().any(axis=1).sum())
```

```python
df_clean = df.copy()

for col in df_clean.columns:
    if df_clean[col].dtype in ["int64", "float64"]:
        df_clean[col] = df_clean[col].fillna(df_clean[col].median())
    elif df_clean[col].dtype == "object":
        df_clean[col] = df_clean[col].fillna(df_clean[col].mode()[0])
    else:
        df_clean[col]= df_clean[col].fillna("unknown")
print (df_clean.isna().sum()) 
```

### Step 5: Visualize Job Satisfaction (Focus on JobSat)


- Create a pie chart or KDE plot to visualize the distribution of `JobSat`.

- Provide an interpretation of the plot, highlighting key trends in job satisfaction.


```python
import matplotlib.pyplot as plt
import seaborn as sns

job_count = df_clean["JobSat"].value_counts()
plt.figure(figsize=(15,10))
plt.pie(job_count, labels= job_count.index, autopct= "%.0f%%", startangle=90, rotatelabels=45,)
plt.title ("Distribution of Job Satisfaction")
plt.show()
```

### Step 6: Programming Languages Analysis


```python
import matplotlib.pyplot as plt
import pandas as pd
langHave = df_clean["LanguageHaveWorkedWith"].dropna().str.split(";").explode().value_counts()
langWant = df_clean["LanguageWantToWorkWith"].dropna().str.split(";").explode().value_counts()
print(langWant.head())
print(langHave.head())

top_languages = list(
    set(langHave.head(10).index) | set(langWant.head(10).index)
)
print("Top Languages", top_languages)

comparison = pd.DataFrame({
    "Currently Used": langHave[top_languages],
    "Want To Work With": langWant[top_languages]
}).fillna(0)

comparison

comparison.plot(
    kind="bar",
    figsize=(12,6)
)

plt.title("Programming Languages: Current Usage vs Future Interest")
plt.xlabel("Programming Language")
plt.ylabel("Number of Developers")
plt.xticks(rotation=45)
plt.show()
```

### Step 7: Analyze Remote Work Trends


- Visualize the distribution of RemoteWork by region using a grouped bar chart or heatmap.

```python
plt.figure(figsize=(10,8))
df_clean["RemoteWork"] = df_clean["RemoteWork"].str.strip()
df_clean["RemoteWork"] = df_clean["RemoteWork"].replace({"Hybrid (some remote, some in-person)":"Hybrid"})
df_clean["Country"] = df_clean["Country"].str.strip()
df_clean["Country"] = df_clean["Country"].replace({"United Kingdom of Great Britain and Northern Ireland": "UK", "United States of America": "US"})
Top_countries = df_clean["Country"].value_counts().head(10).index
Filttered = df_clean[df_clean["Country"]. isin (Top_countries)]

cross = pd.crosstab(Filttered["Country"], Filttered["RemoteWork"])

cross.plot(kind="bar")
plt.title("Distribution of RemoteWork by region/country")
plt.xlabel("Country")
plt.ylabel("Remote")
plt.xticks(rotation =45)
plt.show()

sns.heatmap(cross, annot=True, cmap="Reds", fmt="d")

plt.show()
```

### Step 8: Correlation between Job Satisfaction and Experience


- Analyze the correlation between overall job satisfaction (`JobSat`) and `YearsCodePro`.
  
- Calculate the Pearson or Spearman correlation coefficient.


```python
!pip install scipy
import scipy
df_clean["YearsCodePro"] = df_clean["YearsCodePro"].replace({"Less than 1 year":"0", "More than 50 years":"50", "50+":"50"})
df_clean["YearsCodePro"] = pd.to_numeric(df_clean["YearsCodePro"], errors="coerce")
print(df_clean["JobSat"].dtype)
print(df_clean["YearsCodePro"].dtype)
df_clean["JobSat"].unique()
df_clean["YearsCodePro"].unique()

corr2 = df_clean["JobSat"].corr(
    df_clean["YearsCodePro"],
    method="spearman"
)

print(corr2)

corr = df_clean["JobSat"].corr(
    df_clean["YearsCodePro"],
    method="pearson"
)

print(corr)

```

### Step 9: Cross-tabulation Analysis (Employment vs. Education Level)


- Analyze the relationship between employment status (`Employment`) and education level (`EdLevel`).

- **Instruction**: Create a cross-tabulation using `pd.crosstab()` and visualize it with a stacked bar plot if possible.


```python
top_employment = df_clean["Employment"].value_counts().head().index
print ("TOP EMPLOYMENT", top_employment)
filttered_top = df_clean[df_clean["Employment"].isin(top_employment)]
cross = pd.crosstab(filttered_top["Employment"], filttered_top["EdLevel"]) # normalize="index")
print(cross)
cross.plot(kind="bar", figsize =(12,6))
plt.title("Employment Status by Education Level")
plt.xlabel("Employment Status")
plt.ylabel("Number of Respondents")
plt.xticks(rotation = 45)
plt.legend(title = "Education Level", bbox_to_anchor=(1.05, 1))
plt.show()
```
