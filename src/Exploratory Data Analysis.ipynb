#### Step 4: Analysis of Experience and Job Satisfaction


Analyze the relationship between years of professional coding experience (`YearsCodePro`) and job satisfaction (`JobSat`). Summarize `YearsCodePro` and calculate median satisfaction scores based on experience ranges.

- Create experience ranges for `YearsCodePro` (e.g., `0-5`, `5-10`, `10-20`, `>20` years).

- Calculate the median `JobSat` for each range.

- Visualize the relationship using a bar plot or similar visualization.


```python
# Convert YearsCodePro to numeric
df['YearsCodePro'] = (
    df['YearsCodePro']
    .replace({'Less than 1 year': '0', 'More than 50 years': '50'})
    .astype(float)
)
df['JobSat'] = df['JobSat'].astype(float)
bins = [0, 5, 10, 20, float('inf')]
labels = ['0–5', '5–10', '10–20', '20+']

df['ExpRange'] = pd.cut(df['YearsCodePro'], bins=bins, labels=labels, right=False)
summary = df['YearsCodePro'].describe()
print("Describe:\n", summary)
median_job_sat = df.groupby('ExpRange')['JobSat'].median()
print(median_job_sat)

import seaborn as sns
import matplotlib.pyplot as plt

plt.figure(figsize=(8,5))
sns.barplot(
    data=df,
    x='ExpRange',
    y='JobSat',
    estimator='median',
    order=labels,
    color='skyblue'
)

plt.title('Median Job Satisfaction by Professional Coding Experience')
plt.xlabel('Years of Professional Coding Experience')
plt.ylabel('Median Job Satisfaction')
plt.tight_layout()
plt.show()
```

```python

plt.figure(figsize=(8,5))
sns.barplot(data=median_job_sat)

plt.title('Median Job Satisfaction by Professional Coding Experience')
plt.xlabel('Years of Professional Coding Experience')
plt.ylabel('Median Job Satisfaction')
plt.tight_layout()
plt.show()
```

#### Step 6: Analyzing Remote Work Preferences by Job Role


Analyze trends in remote work based on job roles. Use the `RemoteWork` and `Employment` columns to explore preferences and examine if specific job roles prefer remote work more than others.

- Use a count plot to show remote work distribution.

- Cross-tabulate remote work preferences by employment type (e.g., full-time, part-time) and job roles.


```python
# 1. Count Plot: Remote Work Distribution
# ===========================
plt.figure(figsize=(8,5))
sns.countplot(data=df, x="RemoteWork", palette="viridis")
plt.title("Distribution of Remote Work Preferences")
plt.xlabel("Remote Work Preference")
plt.ylabel("Count")
plt.show()

# ===========================
# 2. Cross-tabulate Remote Work by Employment Type and Job Role
# ===========================

# RemoteWork vs Employment Type
ct_emp_type = pd.crosstab(df["Employment"], df["RemoteWork"], normalize="index") * 100
print("\n=== Remote Work Preference by Employment Type (%) ===")
print(ct_emp_type.head())

# RemoteWork vs Job Role
ct_DevType = pd.crosstab(df["DevType"], df["RemoteWork"], normalize="index") * 100
print("\n=== Remote Work Preference by DevType (%) ===")
print( ct_DevType.head())

# Optional heatmaps
plt.figure(figsize=(10,6))
sns.heatmap(ct_emp_type, annot=True, cmap="Blues", fmt=".1f")
plt.title("Remote Work Preference by Employment Type (%)")
plt.show()

plt.figure(figsize=(12,8))
sns.heatmap(ct_DevType, annot=True, cmap="Greens", fmt=".1f")
plt.title("Remote Work Preference by DevType (%)")
plt.show()
```

#### Step 7: Analyzing Programming Language Trends by Region


Analyze the popularity of programming languages by region. Use the `LanguageHaveWorkedWith` column to investigate which languages are most used in different regions.

- Filter data by country or region.

- Visualize the top programming languages by region with a bar plot or heatmap.


```python
df["LanguageHaveWorkedWith"]
```

```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt


# Ensure needed columns exist
df = df.dropna(subset=["LanguageHaveWorkedWith", "Country"])

# Split languages into rows
df_languages = (
    df.assign(Language=df["LanguageHaveWorkedWith"].str.split(";"))
      .explode("Language")
)

# Clean whitespace
df_languages["Language"] = df_languages["Language"].str.strip()
# Filter by a specific region (optional)
#region_filter = "Europe"
#df_region = df_languages[df_languages["Country"] == region_filter]
lang_counts = (
    df_languages.groupby(["Country", "Language"])
                .size()
                .reset_index(name="Count")
)
print (lang_counts)
```

```python
# Get top N languages per region
top_n = 10
top_by_region = (
    lang_counts.sort_values(["Country", "Count"], ascending=[True, False])
               .groupby("Country")
               .head(top_n)
)
print(top_by_region)
```

```python

plt.figure(figsize=(14, 8))
sns.barplot(
    data=top_by_region,
    x="Count",
    y="Language",
    hue="Country",
    dodge=False
)

plt.title("Top Programming Languages by Country")
plt.xlabel("Number of Users")
plt.ylabel("Language")
plt.legend(title= "Country")
```

#### Step 8: Correlation Between Experience and Satisfaction


Examine how years of experience (`YearsCodePro`) correlate with job satisfaction (`JobSatPoints_1`). Use a scatter plot to visualize this relationship.


```python
# Select relevant columns and drop missing values
df_exp = df[["YearsCodePro", "JobSatPoints_1"]].dropna()

# Convert YearsCodePro to numeric (sometimes contains strings like 'Less than 1 year')
df_exp["YearsCodePro"] = pd.to_numeric(df_exp["YearsCodePro"], errors="coerce")
df_exp = df_exp.dropna()

plt.figure(figsize=(8,6))
sns.scatterplot(
    data=df_exp,
    x="YearsCodePro",
    y="JobSatPoints_1",
    alpha=0.6
)

sns.regplot(
    data=df_exp,
    x="YearsCodePro",
    y="JobSatPoints_1",
    scatter=False,
    color="red"
)

plt.title("Years of Experience vs Job Satisfaction")
plt.xlabel("Years of Professional Coding Experience")
plt.ylabel("Job Satisfaction Score")
plt.show()
```

#### Step 9: Educational Background and Employment Type


Explore how educational background (`EdLevel`) relates to employment type (`Employment`). Use cross-tabulation and visualizations to understand if higher education correlates with specific employment types.


```python
# Keep only the needed columns
df_clean = df[["EdLevel", "Employment"]].dropna()

# Basic cross-tab (counts)
ct = pd.crosstab(df_clean["EdLevel"], df_clean["Employment"])
print("\n=== Cross-tab: EdLevel × Employment (Counts) ===")
print(ct.head(2))

# Percentage version (row percentages)
ct_pct = pd.crosstab(df_clean["EdLevel"], df_clean["Employment"], normalize="index") * 100
print("\n=== Cross-tab: EdLevel × Employment (Percent) ===")
print(ct_pct.round(1).head(2))

```

```python
ppct= ct_pct.head(5)
plt.figure(figsize=(12, 7))
ppct.plot(kind="bar", stacked=False, figsize=(14, 8), colormap="viridis")

plt.title("Employment Type Distribution by Education Level (%)")
plt.ylabel("Percentage")
plt.xlabel("Education Level")
plt.legend(title="Emoyment Typlpe", bbox_to_anchor=(1.05, 1), loc="upper left")
plt.tight_layout()
plt.show()
```

```python
plt.figure(figsize=(15, 10))
sns.heatmap(ct_pct, annot=True, fmt=".1f", cmap="Blues")

plt.title("Heatmap: Education Level vs Employment Type (%)")
plt.xlabel("Employment Type")
plt.ylabel("Education Level")
plt.show()
```
